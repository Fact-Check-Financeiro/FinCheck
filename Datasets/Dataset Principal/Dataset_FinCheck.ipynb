# Dataset do grupo FinCheck

O conjunto de dados foi construído a partir da integração e do processamento de bases de dados preexistentes, contendo notícias classificadas quanto à sua veracidade. A partir dessas bases, foram selecionadas as notícias relacionadas ao domínio financeiro, utilizando técnicas de processamento de linguagem natural e representações vetoriais obtidas por meio do modelo BERTimbau.

Descrição das colunas:
- Noticia: texto das notícias relacionadas à finanças;
- Rotulo: classificação da notícia como "fake" (falsa) ou "true" (verdadeira);
- score_financas: medida de similaridade semântica entre a notícia e as referências relacionadas ao domínio financeiro, definida como o maior valor de similaridade do cosseno obtido entre o embedding da notícia e os embeddings das frases de referência.

Bases de dados utilizadas:
- https://github.com/roneysco/Fake.br-Corpus
- https://huggingface.co/datasets/recogna-nlp/FakeRecogna
